The code below visualizes the model average accuracy tested with cross validation with respect to each dimension.


Note: this can only be run after gaining summary_all.csv and summary_all_SVMRBF.csv under /results directory, which can be obtained by running nn.py and svmrbf.py. However, these two .csv files should be in place and ready for use already.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from pathlib import Path

# data
n = np.array([10, 12, 14, 16, 18, 20])
mlp_acc = pd.read_csv('results/summary_all.csv')['Acc'].values
svm_acc = pd.read_csv('results/summary_all_SVMRBF.csv')['Mean_Acc'].values
target = np.array([0.94, 0.93, 0.92, 0.91, 0.80, 0.75])

# plot
plt.figure(figsize=(6,4))
mlp_acc = pd.read_csv('results/summary_all.csv')['Acc'].values
svm_acc = pd.read_csv('results/summary_all_SVMRBF.csv')['Mean_Acc'].values
plt.plot(n, mlp_acc, 'o-', label='MLP', linewidth=2)
plt.plot(n, svm_acc, 's--', label='SVM-RBF', linewidth=2)
plt.plot(n, target, 'k:', label='Target', linewidth=2)

plt.xlabel('Dimension $n$')
plt.ylabel('CV Accuracy')
plt.title('CV accuracy vs. dimension $n$')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('results/fig_acc_vs_n.png', bbox_inches='tight')
plt.show()
```
