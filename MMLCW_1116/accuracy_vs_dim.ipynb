```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from pathlib import Path

# data
n = np.array([10, 12, 14, 16, 18, 20])
mlp_acc = np.array([0.9606, 0.9539, 0.9587, 0.9315, 0.8749, 0.8609])
svm_acc = np.array([0.954, 0.578, 0.536, 0.514, 0.502, 0.498])
target = np.array([0.94, 0.93, 0.92, 0.91, 0.80, 0.75])

# plot
plt.figure(figsize=(6,4))
plt.plot(n, mlp_acc, 'o-', label='MLP', linewidth=2)
plt.plot(n, svm_acc, 's--', label='SVM-RBF', linewidth=2)
plt.plot(n, target, 'k:', label='Target', linewidth=2)

plt.xlabel('Dimension $n$')
plt.ylabel('CV Accuracy')
plt.title('CV accuracy vs. dimension $n$')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('fig_acc_vs_n.png', bbox_inches='tight')
plt.show()
```
